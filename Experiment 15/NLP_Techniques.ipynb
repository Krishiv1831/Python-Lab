```python
#1. Tokenization

import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize

text = "Ronaldo is the best footballer in the world"

tokens = word_tokenize(text)

print(tokens)
```

```python
# Sentence Tokenization

from nltk.tokenize import sent_tokenize

text = "Messi is the second best footballer in the world."

sentences = sent_tokenize(text)

print(sentences)
```

```python
#Stop word Removal
#Stop words such as is, the, and, in are removed.

from nltk.corpus import stopwords
stopwords = set(stopwords.words('english'))

words = word_tokenize(text)

filtered_words = [w for w in words if w.lower() not in stopwords]

print(filtered_words)
```

```python
#Stemming
#Stemming reduces words to their root form.

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "plays", "played", "running", "watched", "coding"]

for w in words:
    stemmed_word = stemmer.stem(w)
```

```python
#Lemmatization
#Lemmatization converts words to their dictionary form.

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("cats"))
print(lemmatizer.lemmatize("cacti"))
print(lemmatizer.lemmatize("geese"))
print(lemmatizer.lemmatize("rocks"))
```

```python
#Part-of-Speech (POS) Tagging
#POS tagging identifies grammatical role of words.

import nltk
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import pos_tag

words = word_tokenize(text)

pos_tags = pos_tag(words)

print(pos_tags)
```

```python
#Word Frequency Count

from nltk.probability import FreqDist

words = word_tokenize(text)

freq_dist = FreqDist(words)

print(freq_dist.most_common())
```
